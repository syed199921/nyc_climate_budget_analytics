## Stage 1: Business Understanding

### 1.1 Case Study:
New York City (NYC) introduced Climate Budgeting to help decisionmakers monitor how the city’s capital investments support its climate goals. Mayor's Office of Management and Budgeting (OMB) publishes the Climate Alignment Assessment and Capital Climate Investments dataset, which rates capital projects for emissions, flood and heat alignment and records their environmental and social benefits. According to New York City Mayor Eric Adams, the Climate Budgeting initiative aims to improve the health and quality of life of New Yorkers through emissions reductions, enhanced air quality, greater resilience to heat and flooding, and the promotion of environmental justice (Adams, 2024).

#### Sources:
- https://www.nyc.gov/mayors-office/news/2024/04/mayor-adams-celebrates-launch-new-york-city-s-first-climate-budgeting-process

### 1.2 Business Problem

The dataset is a single flat table containing 259,491 rows. It exhibits several data quality and structural issues, including repeated project and budget line details, as well as multiple benefits recorded within a single text field.

These issues make it difficult to answer the following business questions:

#### Objective: Reducing Emissions
- How much planned spending is rated as aligned with emissions reduction, and what proportion of total spending does this represent?
- Which asset categories and budget lines contain the highest levels of spending that are not aligned with emissions reduction?

#### Objective: Enhancing Air Quality
- How much planned funding is allocated to projects that provide an air quality benefit?

#### Objective: Increasing Flood and Heat Resiliency
- How much funding is aligned with heat resiliency, and is it reaching areas identified as high-heat zones?
- How much funding is aligned with flood resiliency, and is it reaching areas identified as high-flood-risk zones?

#### Objective: Social and Environmental Benefits
- How much funding is associated with social and environmental benefits such as access to resources, healthy housing, and reduced hazardous material exposure?

#### Objective: Improving Data Quality and Decision Support
- How much planned spending is still awaiting a climate rating?
- How has the proportion of climate-aligned spending changed across the FY2025, FY2026, and FY2027 plans?

#### Objective: Promoting Environmental Justice Through Heat and Flood Resiliency
- How much funding is allocated to areas that are simultaneously identified as high-heat and high-flood-risk zones?

### 1.3 The Role of Database

The data provide by OMB in the form of a single table could be split into multiple tables such as Projects and Budget Lines. Establishing relationships among them could provide clarity with which the questions mentioned above could be answered.

### 1.4 Project Roadmap

The project includes normalising a MySQL database, performing ETL (Extracting, transformation, and loading) using Python, using SQL to answer the questions mentioned, and exposing the results using an API.
