### 1.1 Case Study:
New York City (NYC) introduced Climate Budgeting to help decisionmakers monitor how the city’s capital investments support its climate goals. Mayor's Office of Management and Budgeting (OMB) publishes the Climate Alignment Assessment and Capital Climate Investments dataset, which rates capital projects for emissions, flood and heat alignment and records their environmental and social benefits. According to New York City Mayor Eric Adams, the Climate Budgeting initiative aims to improve the health and quality of life of New Yorkers through emissions reductions, enhanced air quality, greater resilience to heat and flooding, and the promotion of environmental justice (Adams, 2024).

#### Objective: Social and Environmental Benefits
- How much funding is associated with social and environmental benefits such as access to resources, healthy housing, and reduced hazardous material exposure?

#### Objective: Improving Data Quality and Decision Support
- How much planned spending is still awaiting a climate rating?

#### Objective: Promoting Environmental Justice Through Heat and Flood Resiliency

### Question 4:  Where is the not-aligned and pending spending concentrated?

```python
# Totals check (should match Question 1)
import pandas as pd
not_aligned_ratings = ["Not Aligned", "Not Aligned (Short-Term Benefit)"]
pending_ratings = ["Pending Rating"]

def total_by_plan(ratings):
    return (df[df["ghg_mitigation"].isin(ratings)]
            .groupby("financial_plan")["fiscal_year_amount"].sum())

summary = pd.DataFrame({
    "plan_total": df.groupby("financial_plan")["fiscal_year_amount"].sum(),
    "not_aligned": total_by_plan(not_aligned_ratings),
    "pending": total_by_plan(pending_ratings),
}).fillna(0)
summary["not_aligned_%"] = (summary["not_aligned"] / summary["plan_total"] * 100).round(1)
summary["pending_%"] = (summary["pending"] / summary["plan_total"] * 100).round(1)
print(summary)
```

```python
# Not-aligned money by asset category
def by_group(ratings, group_col):
    part = (df[df["ghg_mitigation"].isin(ratings)]
            .pivot_table(index=group_col, columns="financial_plan",
                         values="fiscal_year_amount", aggfunc="sum", fill_value=0))
    total = df.pivot_table(index=group_col, columns="financial_plan",
                           values="fiscal_year_amount", aggfunc="sum", fill_value=0)
    part = part.reindex(total.index).fillna(0)
    return part, total

na_part, na_total = by_group(not_aligned_ratings, "asset_category")
latest = na_part.columns[-1]   # FY2027 plan, used only for sorting

print("Amount (thousands of US dollars)")
print(na_part.sort_values(latest, ascending=False).to_string())

print("\nShare of each plan's not-aligned money held by the category (%)")
print((na_part / na_part.sum() * 100).round(1)
      .sort_values(latest, ascending=False).to_string())

print("\nNot-aligned share of the category's own total (%)")
print((na_part / na_total.where(na_total != 0) * 100).round(1)
      .loc[na_part.sort_values(latest, ascending=False).index].to_string())
```

```python
# Pending money by asset category
pe_part, pe_total = by_group(pending_ratings, "asset_category")
latest = pe_part.columns[-1]

print("Amount (thousands of US dollars)")
print(pe_part.sort_values(latest, ascending=False).to_string())

print("\nShare of each plan's pending money held by the category (%)")
print((pe_part / pe_part.sum() * 100).round(1)
      .sort_values(latest, ascending=False).to_string())

print("\nPending share of the category's own total (%)")
print((pe_part / pe_total.where(pe_total != 0) * 100).round(1)
      .loc[pe_part.sort_values(latest, ascending=False).index].to_string())
```

```python
# Top ten budget lines (not-aligned and pending)
for name, ratings in [("Not aligned", not_aligned_ratings), ("Pending", pending_ratings)]:
    part, _ = by_group(ratings, "budget_line")
    latest = part.columns[-1]
    print(f"\n{name}: top 10 budget lines by the FY2027 plan (thousands of US dollars)")
    print(part.sort_values(latest, ascending=False).head(10).to_string())
```
